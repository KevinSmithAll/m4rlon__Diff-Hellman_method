# Алгоритм Диффи-Хеллмана

```python
import random
```

```python
A = 'Сторона А'
B = 'Сторона Б'
```

```python
p = 51871 # простое число
g = 10001 # генератор
```

```python
a = random.randint(0, 10_000) # случайное загаданное число стороной А - приватный ключ стороны А
b = random.randint(0, 10_000) # случайное загаданное число стороной Б - приватный ключ стороны Б
```

```python
a, b
```

```python
A_open_key = g**a % p # открытый ключ стороны А
B_open_key = g**b % p  # открытый ключ стороны Б
```

```python
A_open_key
```

```python
B_open_key
```

```python
# обмен ключами
```

### Общий секретный ключ

```python
K_obj_A = B_open_key ** a % p
```

```python
K_obj_B = A_open_key ** b % p
```

```python
K_obj_A == K_obj_B
```
